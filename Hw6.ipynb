W6A16
Nhâ ̣p vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n , m = map(int, input().split())
matrix = []
for i in range(n):
    hang = list(map(int, input().split()))
    matrix.append(hang)
    if len(hang) != m:
        print ("lỗi rồi")
for row in matrix:
    for i in row:
        print(f"{i:4}", end=" ")
    print()
```

Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mản

```python
n = int(input())
matrix = [list(map(int, input().split())) for _ in range(n)]
cheo_chinh = [matrix[i][i] for i in range(n)]
cheo_phu = [matrix[i][n-1-i] for i in range(n)]
print("Đường chéo chính:", cheo_chinh)
print("Đường chéo phụ:", cheo_phu)
```

W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
def anh_xa(list):
    s = input().split()
    d = {}
    for i , word in enumerate(s):
        if word not in d:
            d[word]=[]
            d[word].append(i)
    for word in d:
        d[word]=tuple(d[word])
    return d
```

W6A24
Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí

```python
def xep_so(list):
    so0 = [i for i in list if i == 0]
    conlai = [i for i in list if i != 0]
    return so0 + conlai
```

W6A26
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 đầu cuối 

```python
def tong_dp(list):
    so = list(map(int, input().split()))
    tong = 0
    n = len(so)
    if n <= 3:
        print("Không đủ phần tử")
    else:
        for i in range (1 , n-1):
            if so[i] > so[i-1] and so[i] > so[i+1]:
                tong += so[i]    
    return tong 
```

Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước)

```python
def vi_tri(list, x):
    so = list(map(int, input().split()))
    x = int(input())
    vt = [i for i, gt in enumerate(so) if gt < x]
    if len(vt) == 0:
        print("Không có phần tử nào nhỏ hơn x")
    else:
        print(vt)
```

W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm thực sự hay vô danh 

```python
def ktr_day():
    day = list(map(int, input("Nhập dãy số: ").split()))
    n = len(day)
    
    if n <= 1:
        return "Tang thuc su"
    
    day_tang = all(day[i] < day[i+1] for i in range(n - 1))
    day_giam = all(day[i] > day[i+1] for i in range(n - 1))
    
    if day_tang:
        return "Tang thuc su"
    elif day_giam:
        return "Giam thuc su"
    else:
        return "Vo danh"
```
