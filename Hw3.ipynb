```python
h = int(input("Nhập độ cao h: "))

for i in range(1, h + 1):
    for j in range(h - i):
        print(" ", end="")
    for j in range(1, 2 * i):
        if j == 1 or j == 2 * i - 1 or i == h:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
import math
a = int(input("Nhập a: "))
if a < 0:
    print(f"{a} không phải số chính phương")
else:
    can_bac_hai = math.isqrt(a)
    if can_bac_hai * can_bac_hai == a:
        print(f"{a} là số chính phương")
    else:
        print(f"{a} không phải số chính phương")
```

```python
max= None
min = None

print("Nhập dãy số nguyên (nhập -1 để dừng):")
while True:
    num = int(input("Nhập số: "))
    if num == -1:
        break
    if max is None:
        max = num
        min = num
    else:
        # Cập nhật max và min
        if num > max:
            max = num
        if num < min:
            min = num
if max is not None:
    print(f"Số lớn nhất là: {max}")
    print(f"Số nhỏ nhất là: {min}")
else:
    print("Bạn chưa nhập số nào ngoài số dừng -1.")
```

```python
n = int(input("Nhập n: "))
is_3k = True

if n <= 0:
    is_3k = False
else:
    temp = n
    while temp > 1:
        if temp % 3 != 0:
            is_3k = False
            break
        temp //= 3

if is_3k:
    print(f"{n} có dạng 3^k")
else:
    print(f"{n} không có dạng 3^k")
```

```python
A = int(input("Nhập A: "))
f1, f2 = 1, 1
fib_max = 1

if A < 1:
    print("Không có số Fibonacci nào thỏa mãn.")
else:
    while f2 <= A:
        fib_max = f2
        f1, f2 = f2, f1 + f2
    print(f"Số Fibonacci lớn nhất không vượt quá {A} là: {fib_max}")
```

```python
n = int(input("Nhập n: "))
count = 0
temp = abs(n)
cac_so_nt = [2, 3, 5, 7]

while temp > 0:
    chu_so = temp % 10
    if chu_so in cac_so_nt:
        count += 1
    temp //= 10

print(f"Số lượng chữ số là số nguyên tố: {count}")
```
