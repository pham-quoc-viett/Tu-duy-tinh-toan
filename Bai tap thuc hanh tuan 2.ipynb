```python
import math
a = float(input("Nhập cạnh a: "))
b = float(input("Nhập cạnh b: "))
c = float(input("Nhập cạnh c: "))
if (a + b > c) and (a + c > b) and (b + c > a):
    p = (a + b + c) / 2
    s = math.sqrt(p * (p - a) * (p - b) * (p - c))
    print(f"{s:.1f}")
else:
    print("Khong phai 3 canh cua tam giac")
```

```python
ten = input("Ten chu ho: ")
truoc = int(input("Chi so thang truoc: "))
nay = int(input("Chi so thang nay: "))

dung = nay - truoc
tien = 0

if dung <= 50:
    tien = dung * 1984
elif dung <= 100:
    tien = 50 * 1984 + (dung - 50) * 2050
elif dung <= 200:
    tien = 50 * 1984 + 50 * 2050 + (dung - 100) * 2380
elif dung <= 300:
    tien = 50 * 1984 + 50 * 2050 + 100 * 2380 + (dung - 200) * 2998
elif dung <= 400:
    tien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + (dung - 300) * 3350
else:
    tien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + 100 * 3350 + (dung - 400) * 3460

tong_tien = tien * 1.08
print(f"Ho va ten: {ten}")
print(f"Tien phai tra la: {int(tong_tien)}")
```
