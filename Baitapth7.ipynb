W7A1 : Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target):

```python
def tknp(s, t):
    s1 = sorted(s)
    for i in range (len(s1)):
        if s1[i] == t:
            return i
    return -1
s = list(map(int, input().split()))
t = int(input())
print(tknp(s, t))
```

Cho danh sách arr và một
số nguyên x, đếm xem x
xuất hiện bao nhiêu lần
trong danh sách. def
count_occurrences(arr, x) -
> int:

```python
def dem(s, t):
    count = 0
    for i in range (len(s)):
        if s[i] == t:
            count += 1
    return count
s = list(map(int, input().split()))
t = int(input())
print(dem(s, t))
```

W7A3 : Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort), sắp xếp chọn

```python
# Sắp xếp nổi bọt
def snb(s):
    for i in range (len(s)-1):
        for j in range (len(s)-i-1):
            if s[j] > s[j+1]:
                s[j], s[j+1] = s[j+1], s[j]
    return s
s = list(map(int, input().split())) 
print(snb(s))
```

```python
# SẮP XẾP CHỌN
def sxc(s):
    for i in range (len(s)-1):
        min_index = i
        for j in range (i+1, len(s)):
            if s[j] < s[min_index]:
                min_index = j
        s[i], s[min_index] = s[min_index], s[i]
    return s
s = list(map(int, input().split()))
print(sxc(s))
```

W7A4
Viết hàm tìm phần tử có số
lần xuất hiện nhiều nhất
trong danh sách. Nếu có
nhiều phần tử cùng số lần,
trả về phần tử xuất hiện
trước.

```python
def tkpt(s):
    if not s:
        return None
    s1 = sorted(s)
    max_count = 0
    gt = s1[0]
    for x in (s1):
        count = s1.count(x)
        if count > max_count:
            max_count = count
            gt = x
    return gt
s = list(map(int, input().split()))
print(tkpt(s))

        
```

W7A5
Cho danh sách các số
nguyên và một số X, đếm
tất cả các cặp (a, b) sao
cho a + b = X. c

```python
s = list(map(int, input().split()))
x = int(input())
d=0
for i in range (len(s)):
    for j in range (i+1, len(s)):
        if s[i] + s[j] == x:
            d+= 1
print(d)
```

W7A7
Cho một danh sách các số
nguyên nums, hãy tìm độ
dài của dãy con tăng
nghiêm ngặt dài nhất
(không nhất thiết phải liền kề)

```python
def length_of_lis(nums):
    if not nums:
        return 0
    
    dp = [1] * len(nums)
    
    for i in range(len(nums)):
        for j in range(i):
            if nums[i] > nums[j]:
                if dp[i] < dp[j] + 1:
                    dp[i] = dp[j] + 1
                    
    return max(dp)

nums = list(map(int, input().split()))
print(length_of_lis(nums))
```
