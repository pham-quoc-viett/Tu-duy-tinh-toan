```python
import math
a = float(input())
b = float(input())
c = float(input())
if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        loai = "Tam giác đều"
    elif a == b or b == c or a == c:
        loai = "Tam giác cân"
    elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
        loai = "Tam giác vuông"
    else:
        loai = "Tam giác thường"
    p = (a + b + c) / 2
    s = math.sqrt(p * (p - a) * (p - b) * (p - c))
    print(f"Đây là {loai}. Chu vi: {a+b+c}, Diện tích: {s:.2f}")
else:
    print("Khong la ba canh cua mot tam giac")
```

```python
import math
a = float(input())
b = float(input())
c = float(input())
if a == 0:
    if b == 0:
        print("Vô số nghiệm" if c == 0 else "Vô nghiệm")
    else:
        print(f"x = {-c/b}")
else:
    delta = b**2 - 4*a*c
    if delta < 0:
        print("Phương trình vô nghiệm")
    elif delta == 0:
        print(f"Nghiem kép x = {-b/(2*a)}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)
        print(f"x1 = {x1}, x2 = {x2}")
```

```python
a,b,m,c,d,n = map(float, input().split())
D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c
if D != 0:
    x = Dx / D
    y = Dy / D
    print(f"Hệ có nghiệm duy nhất: x = {x}, y = {y}")
else:
    if Dx == 0 and Dy == 0:
        print("Hệ có vô số nghiệm.")
    else:
        print("Hệ vô nghiệm.")
```

```python
x0, y0, r, x, y = map(float, input().split())
dx = x - x0
dy = y - y0
D = (dx**2 + dy**2)**0.5
if D == r:
    print("Kết luận: Điểm A thuộc đường tròn.")
elif D < r:
    print("Kết luận: Điểm A nằm trong đường tròn.")
else:
    print("Kết luận: Điểm A nằm ngoài đường tròn.")
```
