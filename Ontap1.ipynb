1 Yêu cầu. Cước phí Taxi Thông Minh.
Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả,
kiểu số nguyên, đơn vị VND).
Các quy tắc cần cài đặt:
1. Giá mở cửa (0–0.8 km):
- Xe 4 chỗ: 11,000 VND
- Xe 7 chỗ: 12,000 VND
2. Giá theo km sau 0.8 km:
- Xe 4 chỗ: từ 0.8→20 km ơnh 13,000 VND/km; trên 20 km tính 10,500 VND/km
- Xe 7 chỗ: từ 0.8→20 km ơnh 14,000 VND/km; trên 20 km tính 12,000 VND/km
3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND.
4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường +
chờ).
5. Điều kiện đặc biệt:
- Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền.
- Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu).

```python
import math

def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    fare_distance = 0
    if car_type == 4:
        fare_distance += 11000
        if km > 0.8:
            if km <= 20:
                fare_distance += (km - 0.8) * 13000
            else:
                fare_distance += (20 - 0.8) * 13000 + (km - 20) * 10500
    elif car_type == 7:
        fare_distance += 12000
        if km > 0.8:
            if km <= 20:
                fare_distance += (km - 0.8) * 14000
            else:
                fare_distance += (20 - 0.8) * 14000 + (km - 20) * 12000
    wait_fare = 0
    if wait_minutes > 5:
        blocks = math.ceil((wait_minutes - 5) / 5)
        wait_fare = blocks * 2000
    
    total = fare_distance + wait_fare
    is_night = start_hour in [22, 23, 0, 1, 2, 3, 4, 5]
    if is_night:
        total *= 1.2
    if km > 30 and not is_night:
        total *= 0.9
    if total < 50000 and km > 5:
        total = 50000

    return int(total)
```

4 Yêu cầu: Cho các hệ số thực a, b, c, trả về tất cả các nghiệm thực của phương trình bậc hai axx + bx + c = 0 trong một danh
sách được sắp xếp tăng dần.
Quy tắc:
- Nếu a == 0 và b != 0: trả về nghiệm duy nhất [-c / b].
- Nếu a == 0 và b == 0: trả về [] (không có nghiệm hữu hạn duy nhất nếu c == 0, hoặc vô nghiệm nếu c != 0).
- Với a != 0, dùng biệt thức d = bb - 4ac:
- d < 0 → []
- d == 0 → [ -b / (2a) ]
- d > 0 → hai nghiệm [r1, r2] với r1 <= r2

```python
def gpt(a, b, c):
    if a == 0:
        if b == 0:
            return []
        return [-c / b]
    
    delta = b**2 - 4*a*c
    if delta < 0:
        return []
    elif delta == 0:
        return [-b / (2*a)]
    else:
        x1 = (-b - delta**0.5) / (2*a)
        x2 = (-b + delta**0.5) / (2*a)
        return sorted([x1, x2])
```

16 Viết hàm trả về góc ∠ABC theo đơn vị độ khi biết tọa độ các điểm A(x1,y1), B(x2,y2), C(x3,y3). 

```python
def goc_abc(x1, y1, x2, y2, x3, y3):
    ba = (x1 - x2, y1 - y2)
    bc = (x3 - x2, y3 - y2)
    dot_product = ba[0]*bc[0] + ba[1]*bc[1]
    ma = math.sqrt(ba[0]**2 + ba[1]**2)
    mc = math.sqrt(bc[0]**2 + bc[1]**2)
    cos_angle = dot_product / (ma * mc)
    return math.degrees(math.acos(max(-1, min(1, cos_angle))))
```

và False nếu ngược lại.
- Đầu vào:
- quad: danh sách hoặc bộ gồm đúng 4 đỉnh [(x1,y1), (x2,y2), (x3,y3), (x4,y4)] theo thứ tự (tứ giác có thể lồi hoặc lõm,
nhưng không tự cắt).
- p: một điểm (x, y).
- Đầu ra: giá trị boolean.
- Gợi ý: Cài đặt phép kiểm tra điểm trong đa giác tổng quát (ví dụ: ray casting), đồng thời tính cả trường hợp điểm nằm
trên biên là ở trong. 


```python
def point_in_quadrilateral(quad, p):
    x, y = p
    n = len(quad)
    inside = False
    
    for i in range(n):
        p1 = quad[i]
        p2 = quad[(i + 1) % n]
        if p1[1] == p2[1] == y:
            if min(p1[0], p2[0]) <= x <= max(p1[0], p2[0]):
                return True
        if p1[0] == p2[0] == x:
            if min(p1[1], p2[1]) <= y <= max(p1[1], p2[1]):
                return True
        if ((p1[1] > y) != (p2[1] > y)) and \
           (x < (p2[0] - p1[0]) * (y - p1[1]) / (p2[1] - p1[1]) + p1[0]):
            inside = not inside
    return inside
```

26 Yêu cầu: Viết hàm are_collinear(x1, y1, x2, y2, x3, y3) trả về True nếu ba điểm thẳng hàng, ngược lại trả về False.
- Đầu vào: ba điểm x1, y1, x2, y2, x3, y3 biểu diễn tọa độ của ba điểm.
- Đầu ra: giá trị boolean.
- Gợi ý: Dùng phép kiểm tra diện tích (tích có hướng) và một sai số nhỏ khi so sánh số thực. Các điểm nằm trên cùng một
đường thẳng được xem là thẳng hàng (kể cả khi có điểm trùng nhau).

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    val = x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)
    epsilon = 1e-9
    return abs(val) < epsilon
```

- Đầu vào: các số thực a, b, c.

- Nếu a == 0 và b == 0 và c != 0, phương trình vô nghiệm. 

```python
def quadratic_roots_count(a, b, c):
    if a == 0:
        if b == 0:
            if c == 0:
                return float('inf')
            else:
                return 0 
        else:
            return 1
    delta = b**2 - 4*a*c
    
    if delta > 0:
        return 2
    elif delta == 0:
        return 1
    else:
        return 0
```

30 Yêu cầu: Viết hàm detect_number_type(s) trả về 'int' nếu s biểu diễn một số nguyên, 'float' nếu biểu diễn một số thực dấu
chấm động, ngược lại trả về 'neither'.
- Đầu vào: chuỗi s.
- Đầu ra: một trong các giá trị {'int', 'float', 'neither'}.
- Quy tắc:
- Cho phép có dấu + hoặc - ở đầu.
- Số nguyên phải chỉ gồm các chữ số và không có dấu chấm thập phân (ví dụ: '100').
- Số thực phải có dấu chấm và có chữ số ở cả hai phía của dấu chấm (ví dụ: '100.0', '-3.14').
- Các chuỗi như '.5', '5.', '1e3' được xem là 'neither' trong bài này. 

```python
def detect_number_type(s):
    temp = s
    if s.startswith('+') or s.startswith('-'):
        temp = s[1:]
    if not temp:
        return 'neither'
    if temp.isdigit():
        return 'int'
    if temp.count('.') == 1:
        parts = temp.split('.')
        if parts[0].isdigit() and parts[1].isdigit():
            return 'float'

    return 'neither'
```

32 Yêu cầu: Viết hàm is_right_triangle(a, b, c) trả về True nếu a, b, c có thể là ba cạnh của một tam giác vuông, ngược lại trả
về False.
- Đầu vào: các số dương a, b, c.
- Đầu ra: giá trị boolean.
- Lưu ý: Kiểm tra bất đẳng thức tam giác và định lý Pythagore. Dùng sai số nhỏ khi so sánh số thực.

```python
def is_right_triangle(a, b, c):
    if not (a + b > c and a + c > b + b + c > a):
        return False
    sides = sorted([a, b, c])
    s1, s2, h = sides[0], sides[1], sides[2]
    epsilon = 1e-9
    return abs(s1**2 + s2**2 - h**2) < epsilon
```

33 Yêu cầu: Viết hàm compute(a, b, op) tính a (+|-||/) b tùy theo op.
- Đầu vào: các số a, b và ký tự op thuộc {'+', '-', '', '/'}.
- Đầu ra: kết quả số học.
- Yêu cầu thêm:
- Nếu op là '/' và b == 0, phát sinh ValueError.
- Nếu op không hợp lệ, phát sinh ValueError.

```python
def compute(a, b, op):
    if op == '+':
        return a + b
    elif op == '-':
        return a - b
    elif op == '*':
        return a * b
    elif op == '/':
        if b == 0:
            raise ValueError("Không thể chia cho 0.")
        return a / b
    else:
        raise ValueError("Toán tử không hợp lệ.")
```

35 Yêu cầu: Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) trong đó q là thương nguyên và r là số dư sao
cho dividend = divisorq + r với 0 <= |r| < |divisor| (dùng ngữ nghĩa của Python với // và %).
- Đầu vào: số nguyên dividend, divisor (divisor khác 0).
- Đầu ra: bộ hai số nguyên (q, r).
- Ví dụ: divide_with_remainder(5, 3) -> (1, 2). 

```python
def divide_with_remainder(dividend, divisor):
    if divisor == 0:
        raise ValueError("Số chia phải khác 0.")
    
    q = dividend // divisor
    r = dividend % divisor
    return (q, r)
```

40 Viết hàm chuyển một chữ cái in hoa thành chữ thường. Nếu đầu vào không phải đúng một ký tự, phát sinh ValueError.

```python
def to_lower(c):
    if len(c) != 1:
        raise ValueError("Đầu vào phải là một ký tự duy nhất.")
    return c.lower()
```

41 Viết hàm trả về ký tự đứng sau ký tự x n vị trí trong bảng chữ cái.
Giữ nguyên chữ hoa/chữ thường và quay vòng trong phạm vi A..Z hoặc a..z. 

```python
def shift_char(char, n):
    if 'a' <= char <= 'z':
        return chr((ord(char) - ord('a') + n) % 26 + ord('a'))
    elif 'A' <= char <= 'Z':
        return chr((ord(char) - ord('A') + n) % 26 + ord('A'))
    return char
```

43 Viết hàm tính diện tích tam giác có ba cạnh a, b, c (dùng công thức Heron). Phát sinh ValueError nếu ba cạnh không thể tạo thành tam giác.

```python
def triangle_area_heron(a, b, c):
    if not is_triangle(a, b, c):
        raise ValueError("Ba cạnh không tạo thành tam giác.")
    s = (a + b + c) / 2
    area = math.sqrt(s * (s - a) * (s - b) * (s - c))
    return area
```

44 Viết hàm trả về tích các số nguyên từ 1 đến n (giai thừa). Với n=0, trả về 1.

```python
def gt(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
```

45 Viết hàm trả về số lượng số nguyên dương k sao cho 1 <= k < n và k > (n - 1) / 2. (Ví dụ: n=9 -> 4)

```python
def count_k(n):
    dem = 0
    s = (n - 1) / 2
    for k in range(1, n):
        if k > s:
            dem += 1
    return dem
```
