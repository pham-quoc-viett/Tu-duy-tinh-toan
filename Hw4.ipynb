```python
n = int(input("Nhập n: "))
temp = n
if n <= 0:
    print(f"{n} không phải dạng 2^k")
else:
    while n % 2 == 0:
        n //= 2
    if n == 1:
        print(f"{temp} là số dạng 2^k")
    else:
        print(f"{temp} không phải dạng 2^k")
```

```python
n = int(input("Nhập n: "))
chan = 0
le = 0
while n > 0:
    chu_so = n % 10
    if chu_so % 2 == 0:
        chan += 1
    else:
        le += 1
    n //= 10
print(f"Số chữ số chẵn: {chan}, số chữ số lẻ: {le}")
```

```python
max_val = None
min_val = None

while True:
    num = int(input("Nhập số (-1 để dừng): "))
    if num == -1:
        break
    if max_val is None or num > max_val:
        max_val = num
    if min_val is None or num < min_val:
        min_val = num

print(f"Max: {max_val}, Min: {min_val}")
```

```python
a = int(input("Nhập A: "))
f1, f2 = 1, 1
while f1 + f2 <= a:
    f_next = f1 + f2
    f1 = f2
    f2 = f_next
print(f"Số Fibonacci lớn nhất không quá {a} là: {f2}")
```

```python
s = input()
hoa = 0
thuong = 0
so = 0
for char in s:
    if char.isupper():
        hoa += 1
    elif char.islower():
        thuong += 1
    elif char.isdigit():
        so += 1
print("Ký tự in hoa:", hoa)
print("Ký tự in thường:", thuong)
print("Ký tự số:", so)
```
