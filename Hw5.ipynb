w5a16 Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

```python
def is_armstrong(x):
    if x < 0:
        return False
    s = str(x)
    n = len(s)
    total = sum(int(digit)**n for digit in s)
    return total == x
```

w5a17 Viết hàm kiểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

```python
import math
def is_perfect_square(x):
    if x < 0:
        return False
    root = int(math.isqrt(x))
    return root * root == x
```

w5a19 VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.

```python
def is_magic_number(x):
    while x > 9:
        total = 0
        s = str(x)
        for char in s:
            digit = int(char)
            total += digit
        x = total
    return x == 1
```

w5a20 Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

```python
def detect_number_type(s):
    temp = s.lstrip('-')
    
    if temp.isdigit():
        return 'int'
    
    if '.' in temp:
        parts = temp.split('.')
        if len(parts) == 2:
            combined = parts[0] + parts[1]
            if combined.isdigit() and len(combined) > 0:
                return 'float'
                
    return 'neither'
```

```python
def quadratic_roots_count(a, b, c):
    if a == 0:
        if b == 0:
            return 0 if c != 0 else float('inf')
        return 1
    delta = b**2 - 4*a*c
    if delta > 0:
        return 2
    elif delta == 0:
        return 1
    return 0
```

w5a24 Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (y2 - y1) * (x3 - x2) == (y3 - y2) * (x2 - x1)
```

w5a25 Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)

```python
def point_in_quadrilateral(quad, p):
    import math
    def get_area(p1, p2, p3):
        return abs(p1[0]*(p2[1] - p3[1]) + p2[0]*(p3[1] - p1[1]) + p3[0]*(p1[1] - p2[1])) / 2.0
    q1, q2, q3, q4 = quad
    area_quad = get_area(q1, q2, q3) + get_area(q1, q3, q4)
    area_p = (get_area(p, q1, q2) + get_area(p, q2, q3) + 
              get_area(p, q3, q4) + get_area(p, q4, q1))
    return math.isclose(area_quad, area_p)
```

w5a27 Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)

```python
def count_non_zero_digits(lst):
    total_count = 0
    for n in lst:
        for digit in str(n):
            if digit != '0' and digit.isdigit():
                total_count += 1
    return total_count
```

w5a28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).

```python
import math
def calculate_angle(x1, y1, x2, y2, x3, y3):
    ba = (x1 - x2, y1 - y2)
    bc = (x3 - x2, y3 - y2)
    dot_prod = ba[0]*bc[0] + ba[1]*bc[1]
    mag_ba = math.sqrt(ba[0]**2 + ba[1]**2)
    mag_bc = math.sqrt(bc[0]**2 + bc[1]**2)
    res = math.acos(dot_prod / (mag_ba * mag_bc))
    return math.degrees(res)
```

w5a33 Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa, giá km, phí chờ, phụ phí ban đêm và giảm giá.

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4:
        fare = 11000 + max(0, km - 0.8) * 15000
    else:
        fare = 12000 + max(0, km - 0.8) * 17000
    fare += wait_minutes * 500
    if 22 <= start_hour or start_hour <= 5:
        fare *= 1.1
    return fare
```

w5a34 Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau
Nếu n ≤ 0 trả về 0

```python
def sum_1_to_n(n):
    if n <= 0:
        return 0
    total = 0
    for i in range(1, n + 1):
        s = str(i)
        if len(s) == len(set(s)):
            total += i
    return total
```
