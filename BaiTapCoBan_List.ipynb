### Bài 1: Tìm số lớn nhất của 1 list. KHONG DUNG HAM SAN CO CUA LIST

```python
def tim_max(a):
    max_val = a[0]
    for x in a:
        if x > max_val:
            max_val = x
    return max_val

lst = list(map(int, input().split()))
print(tim_max(lst))
```

### Bài 2: Tìm số nhỏ nhất của một danh sách các số nguyên. KHONG DUNG HAM SAN CO CUA LIST

```python
def tim_min(a):
    min_val = a[0]
    for x in a:
        if x < min_val:
            min_val = x
    return min_val

lst = list(map(int, input().split()))
print(tim_min(lst))
```

### Bài 2b: Tìm số nhỏ nhất của 1 list và vị trí của nó, nếu có nhiều số nhỏ nhất bằng nhau thì lấy chỉ số đầu tiên (nhỏ nhất)
10 0 0 12 34 5 6 7  8 89 0 1
0 1

```python

def tim_min_va_vi_tri(a):
    min_val = a[0]
    index = 0
    for i in range(len(a)):
        if a[i] < min_val:
            min_val = a[i]
            index = i
    return min_val, index

lst = list(map(int, input().split()))
val, pos = tim_min_va_vi_tri(lst)
print(val, pos)
```

### Bài 3: 
Tìm giá trị trung bình của các số nguyên dương trong một list.
Nếu không có số dương nào thì trả về "invalid"

```python
def trung_binh_duong(a):
    tong = 0
    dem = 0
    for x in a:
        if x > 0:
            tong = tong + x
            dem = dem + 1
    if dem > 0:
        return tong / dem
    return "invalid"

lst = list(map(int, input().split()))
print(trung_binh_duong(lst))
```

### Bài 4: Kiểm tra một list số nguyên có phải là đối xứng hay không
[1, 2, 3, 5, 3, 2, 1]

```python
def kiem_tra_doi_xung(a):
    n = len(a)
    for i in range(n // 2):
        if a[i] != a[n - 1 - i]:
            return False
    return True

lst = [1, 2, 3, 5, 3, 2, 1]
print(kiem_tra_doi_xung(lst))
```

### Bài 5: Đưa tất cả các số bằng 0 trong list về đầu của list. Các số khác không thay đổi thứ tự trước sau
3 4 6 7 1 9 0 0 23 0 0 34 67 89
0 0 0 0 3 4 6 7 1 9 23 34 67 89

```python
def dua_0_ve_dau(a):
    kq = []
    for x in a:
        if x == 0:
            kq = [0] + kq
        else:
            kq = kq + [x]
    return kq

lst = [3, 4, 6, 7, 1, 9, 0, 0, 23, 0, 0, 34, 67, 89]
print(dua_0_ve_dau(lst))
```

### Bài 6: Xóa tất cả các số giống nhau trong list, giữ lại đúng 1 lần, Các số khác giữ nguyên vị trí trước 
[1, 2, 3, 4, 1, 2, 3, 4, 1, 1, 4, 3] 


```python
def xoa_trung(a):
    kq = []
    for x in a:
        da_co = False
        for y in kq:
            if x == y:
                da_co = True
                break
        if not da_co:
            kq = kq + [x]
    return kq

lst = list(map(int, input().split()))
print(xoa_trung(lst))
```

### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang

9 8 7 6 5 4 0: Giam

1 2 3 7 1 0 2: Vo danh

1 2 2 4 5 7: Vo danh

```python
def kiem_tra_day(a):
    tang = True
    giam = True
    for i in range(len(a) - 1):
        if a[i] >= a[i + 1]:
            tang = False
        if a[i] <= a[i + 1]:
            giam = False
            
    if tang:
        return "Tang"
    if giam:
        return "Giam"
    return "Vo danh"

s = list(map(int, input().split()))
print(kiem_tra_day(s))
```

### Bài 8: Kiểm tra một danh sách số nguyên có phải là ziczac khong
1 3 2 5 2 8 7 9 2 100: ziczac
34 2 56 7 90 20 30 7 120: ziczac
2 2 34 4 56 7 89 : không phải ziczac


```python
def kiem_tra_ziczac(a):
    n = len(a)
    if n < 3:
        return "ziczac"
    for i in range(1, n - 1):
        is_up = a[i] > a[i-1] and a[i] > a[i+1]
        is_down = a[i] < a[i-1] and a[i] < a[i+1]
        if not (is_up or is_down):
            return "khong phai ziczac"
    return "ziczac"

s = list(map(int, input().split()))
print(kiem_tra_ziczac(s))
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước. Nếu không có k trong ds thì in ra "-1".
VD: k = 3, [1,3,4,4,3,3,4,3,6,1] => k xuất hiện tại các vị trí (trả về 1 list) [1, 4, 5, 7]
k = 10, [1,3,4,4,3,3,4,3,6,1] => k không xuất hiện trong ds => trả về None

```python
def tim_vi_tri_k(a, k):
    kq = []
    for i in range(len(a)):
        if a[i] == k:
            kq = kq + [i]
    if len(kq) == 0:
        print("-1")
        return None
    return kq

s = list(map(int, input().split()))
k = int(input())
print(tim_vi_tri_k(s, k))
```

### Bài 11: 
Cho một danh sách các số nguyên có giá trị trong đoạn [0,9]. Liệt kê số lần xuất hiện của các số có mặt trong danh sách. Số nào xuất hiện trước thì liệt kê trước
VD 1 2 1 2 1 1 1 1

output: 
1: 6
2: 2


```python
def dem_theo_thu_tu(a):
    da_duyet = []
    for x in a:
        check = False
        for d in da_duyet:
            if x == d:
                check = True
                break
        if not check:
            dem = 0
            for y in a:
                if x == y:
                    dem = dem + 1
            print(str(x) + ": " + str(dem), end=" ")
            da_duyet = da_duyet + [x]

s = list(map(int, input().split()))
dem_theo_thu_tu(s)
```

### Bài 11B: 
Cho một danh sách các số nguyên có giá trị trong đoạn [0,9]. Liệt kê số lần xuất hiện của các số có mặt trong danh sách. Số nào nhỏ thì liệt kê trước
VD 2 1 2 1 2 1 1 1 1

output: dict 



```python
def dem_theo_gia_tri(a):
    kq = {}
    for i in range(10):
        dem = 0
        for x in a:
            if x == i:
                dem = dem + 1
        if dem > 0:
            kq[i] = dem
    return kq

lst = list(map(int, input().split()))
print(dem_theo_gia_tri(lst))
```

### Bài 12: 
Cho một chuỗi kí tự chỉ gồm các kí tự chữ cái thường 'a'....'z'. Hãy liệt kê số lần xuất hiện của các kí tự có mặt trong chuỗi theo thứ tự tăng dần.
VD: "abcabcccc"
output:
a : 2
b : 2
c : 5

```python
def dem_ky_tu_alphabet(s):
    bang_chu_cai = "abcdefghijklmnopqrstuvwxyz"
    for chu in bang_chu_cai:
        dem = 0
        for ky_tu in s:
            if ky_tu == chu:
                dem = dem + 1
        if dem > 0:
            print(chu + " : " + str(dem), end=" ")

chuoi = input()
dem_ky_tu_alphabet(chuoi)
```
